# CellIQ CT nodule detector: exploration
Downloads the MONAI lung nodule detection bundle and prints its files so we can wire it up correctly. No big data yet.

```python
import subprocess, sys
r = subprocess.run([sys.executable, "-m", "pip", "install", "-q", "monai[itk,nibabel,tqdm]", "fire", "pytorch-ignite"],
                   capture_output=True, text=True)
print("pip exit code:", r.returncode)
print(r.stderr[-600:] if r.returncode else "ok")
import monai, torch
print("monai", monai.__version__, "| torch", torch.__version__, "| cuda:", torch.cuda.is_available())
```

```python
r = subprocess.run([sys.executable, "-m", "monai.bundle", "download", "lung_nodule_ct_detection", "--bundle_dir", "/content/bundles"],
                   capture_output=True, text=True)
print("download exit code:", r.returncode)
print((r.stdout + r.stderr)[-1500:])
```

```python
import os
root = "/content/bundles/lung_nodule_ct_detection"
for dp, dn, fn in os.walk(root):
    for f in sorted(fn):
        p = os.path.join(dp, f)
        print(os.path.relpath(p, root), os.path.getsize(p))
```

```python
import json
meta = json.load(open(root + "/configs/metadata.json"))
print("name:", meta.get("name"), "| version:", meta.get("version"))
print("license/copyright:", meta.get("copyright"), "|", meta.get("authors"))
print("monai_version:", meta.get("monai_version"), "| pytorch:", meta.get("pytorch_version"))
print("task:", meta.get("task"))
print("description:", (meta.get("description") or "")[:600])
print("image_classes:", json.dumps(meta.get("network_data_format", {}), indent=1)[:1800])
```
